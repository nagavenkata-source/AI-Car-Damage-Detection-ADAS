```python
# ── Improved analyzer: CSV + Annotated JPG + PDF + (optional) S3 upload ─────────
import io, os, csv, json
from datetime import datetime, timezone
from PIL import Image, ImageDraw
import boto3

# Load AWS clients from config if not present already
try:
    REGION
    MODEL_ARN
except NameError:
    with open("config_aws.json", "r") as f:
        _CFG = json.load(f)
    REGION    = _CFG.get("region", "us-east-2")
    MODEL_ARN = _CFG["model_arn"]
    S3_BUCKET = _CFG.get("s3_bucket", "")
    rek = boto3.client("rekognition", region_name=REGION)
    s3  = boto3.client("s3", region_name=REGION) if S3_BUCKET else None

# If you renamed the PDF helper, set this to your function:
#  - make_damage_report_pdf(...)  (professional)
#  - or make_pdf_from_annotated(..., label, conf)
PDF_HELPER_NAME = "make_damage_report_pdf"

def _upload_to_s3(file_path, key_prefix="reports/", expires_sec=86400):
    """Upload a file to S3 and return a presigned URL (or None if bucket missing)."""
    if not S3_BUCKET or s3 is None:
        return None
    key = key_prefix + os.path.basename(file_path)
    s3.upload_file(file_path, S3_BUCKET, key)
    return s3.generate_presigned_url(
        "get_object",
        Params={"Bucket": S3_BUCKET, "Key": key},
        ExpiresIn=expires_sec,
    )

def analyze_severity(image_path, min_conf=20.0, out_dir="rek_results", upload=False):
    os.makedirs(out_dir, exist_ok=True)

    # 1) call Rekognition
    with open(image_path, "rb") as f:
        img_bytes = f.read()

    resp = rek.detect_custom_labels(
        ProjectVersionArn=MODEL_ARN,
        Image={"Bytes": img_bytes},
        MinConfidence=min_conf
    )

    labels = sorted(resp.get("CustomLabels", []),
                    key=lambda x: x.get("Confidence", 0), reverse=True)
    top_name = labels[0]["Name"] if labels else "NO_LABEL"
    top_conf = float(labels[0].get("Confidence", 0.0)) if labels else 0.0
    print(f"Top severity: {top_name}  (confidence: {top_conf:.2f}%)")
    if len(labels) > 1:
        print("All labels:", [(l['Name'], round(l['Confidence'],2)) for l in labels])

    # 2) annotate banner (+ boxes if present)
    im = Image.open(io.BytesIO(img_bytes)).convert("RGB")
    draw = ImageDraw.Draw(im)
    banner_h = max(40, im.height // 16)
    color_map = {"minor": (255,193,7), "moderate": (255,87,34), "severe": (220,53,69)}
    banner_color = color_map.get(top_name.lower(), (0,123,255))
    draw.rectangle([(0,0),(im.width, banner_h)], fill=banner_color)
    draw.text((10, banner_h//4), f"{top_name.upper()} ({top_conf:.1f}%)", fill=(255,255,255))

    for lab in labels:
        box = lab.get("Geometry", {}).get("BoundingBox")
        if box:
            L = int(box["Left"] * im.width); T = int(box["Top"] * im.height)
            W = int(box["Width"] * im.width); H = int(box["Height"] * im.height)
            draw.rectangle([L, T, L+W, T+H], outline=(0,255,0), width=3)
            draw.text((L, max(0, T-18)), f"{lab['Name']} {lab['Confidence']:.1f}%", fill=(255,255,255))

    base = os.path.splitext(os.path.basename(image_path))[0]
    out_img = os.path.join(out_dir, f"{base}_annotated.jpg")
    im.save(out_img, quality=92)

    # 3) append CSV (timezone-aware)
    log_csv = os.path.join(out_dir, "damage_severity_log.csv")
    new = not os.path.exists(log_csv)
    with open(log_csv, "a", newline="") as f:
        w = csv.writer(f)
        if new:
            w.writerow(["timestamp_utc","image","top_label","top_confidence","all_labels"])
        w.writerow([
            datetime.now(timezone.utc).isoformat(),
            image_path,
            top_name,
            round(top_conf,2),
            "; ".join([f"{l['Name']}:{round(l['Confidence'],2)}" for l in labels])
        ])

    # 4) make PDF (use your professional helper if available)
    if PDF_HELPER_NAME == "make_damage_report_pdf" and "make_damage_report_pdf" in globals():
        pdf_path = make_damage_report_pdf(out_img, top_name, top_conf, out_dir=out_dir)
    elif "make_pdf_from_annotated" in globals():
        pdf_path = make_pdf_from_annotated(out_img, top_name, top_conf, out_dir=out_dir)
    else:
        pdf_path = None  # no helper defined — skip quietly

    # 5) optional S3 upload
    s3_links = {}
    if upload:
        if out_img:
            s3_links["annotated_url"] = _upload_to_s3(out_img)
        if pdf_path:
            s3_links["pdf_url"] = _upload_to_s3(pdf_path)

    return {
        "top_label": top_name,
        "confidence": top_conf,
        "out_image": out_img,
        "log_csv": log_csv,
        "pdf_report": pdf_path,
        "s3_links": s3_links
    }
```

```python
# --- PDF helper: builds a 1-page PDF with title + timestamp + annotated image ---
from datetime import datetime, timezone
from PIL import Image, ImageDraw
import os

def make_pdf_from_annotated(annotated_path, label, conf, out_dir="rek_results"):
    os.makedirs(out_dir, exist_ok=True)
    base = os.path.splitext(os.path.basename(annotated_path))[0].replace("_annotated","")
    out_pdf = os.path.join(out_dir, f"{base}_report.pdf")

    # A4-ish page ~150dpi
    PAGE_W, PAGE_H = 1240, 1754
    M = 60

    with Image.new("RGB", (PAGE_W, PAGE_H), (255,255,255)) as page:
        draw = ImageDraw.Draw(page)
        draw.text((M, M), "Damage Report", fill=(0,0,0))
        draw.text((M, M+40), f"Detected: {label.upper()} ({conf:.1f}%)", fill=(0,0,0))
        draw.text((M, M+80), f"Timestamp (UTC): {datetime.now(timezone.utc).isoformat()}", fill=(0,0,0))

        with Image.open(annotated_path).convert("RGB") as img:
            img.thumbnail((PAGE_W-2*M, PAGE_H-(M+140)-M))
            x = (PAGE_W - img.width)//2
            y = M + 140
            page.paste(img, (x, y))

        page.save(out_pdf, "PDF", resolution=300.0)

    return out_pdf
```

```python
from datetime import datetime, timezone
from PIL import Image, ImageDraw, ImageFont
import os, textwrap

def make_damage_report_pdf(
    annotated_path, label, conf, *,
    out_dir="rek_results",
    project_title="AI-Driven Car Damage Detection + ADAS",
    student="Chaitanya",
    university="Gannon University",
    course="MSc – Computer Information Science",
    model_note=None
):
    # -------- setup --------
    os.makedirs(out_dir, exist_ok=True)
    base = os.path.splitext(os.path.basename(annotated_path))[0].replace("_annotated","")
    out_pdf = os.path.join(out_dir, f"{base}_report.pdf")

    # page & colors
    W, H = 1240, 1754
    pad = 60
    WHITE = (255,255,255)
    BLUE  = (32,107,196)
    BLACK = (20,20,20)
    GRAY  = (230,233,236)
    ACCENT = {"minor": (255,193,7), "moderate": (255,87,34), "severe": (220,53,69)}.get(label.lower(), BLUE)

    # fonts
    def _try_font(size):
        for p in [
            "/Library/Fonts/Arial.ttf",
            "/System/Library/Fonts/Supplemental/Arial.ttf",
            "/System/Library/Fonts/Helvetica.ttc",
        ]:
            if os.path.exists(p):
                try:
                    return ImageFont.truetype(p, size)
                except Exception:
                    pass
        return ImageFont.load_default()

    title_font = _try_font(44)
    sub_font   = _try_font(24)
    body_font  = _try_font(22)
    small_font = _try_font(20)

    # -------- canvas --------
    page = Image.new("RGB", (W, H), WHITE)
    draw = ImageDraw.Draw(page)

    # --- header
    header_h = 110
    draw.rectangle([0,0,W,header_h], fill=BLUE)
    draw.text((pad, 28), "Vehicle Damage Assessment Report", fill=WHITE, font=title_font)
    draw.text((pad, 72), project_title, fill=(235,242,255), font=sub_font)

    # --- metadata card (left)
    col_gap = 40
    left_w  = 480
    right_w = W - (pad*2 + col_gap + left_w)

    card_x = pad
    card_y = header_h + pad
    card_h = H - header_h - pad*2 - 50
    draw.rounded_rectangle([card_x, card_y, card_x+left_w, card_y+card_h], radius=20, fill=GRAY)

    # label pill
    pill_h = 48
    draw.rounded_rectangle([card_x+24, card_y+24, card_x+24+left_w-48, card_y+24+pill_h],
                           radius=16, fill=ACCENT)
    draw.text((card_x+40, card_y+28), f"{label.upper()}  ({conf:.1f}%)",
              fill=BLACK if label.lower()=="minor" else WHITE, font=body_font)

    # info lines
    line_y = card_y + 24 + pill_h + 28
    info = {
        "Timestamp (UTC)": datetime.now(timezone.utc).strftime("%Y-%m-%d %H:%M:%S"),
        "Source file": os.path.basename(annotated_path).replace("_annotated",""),
        "Student": student,
        "Course": course,
        "University": university,
    }
    if model_note:
        info["Model"] = model_note

    for k,v in info.items():
        draw.text((card_x+32, line_y), f"{k}:", fill=BLACK, font=body_font)
        wrapped = textwrap.fill(str(v), width=32)
        draw.text((card_x+32, line_y+30), wrapped, fill=(60,60,60), font=small_font)
        line_y += 90

    # --- annotated image (right) — BIG
    with Image.open(annotated_path).convert("RGB") as img:
        # allow almost full page height minus header/footer, and full right column width
        usable_h = H - (header_h + pad*3)
        usable_w = W - (left_w + col_gap + pad*2)

        # keep aspect ratio & scale aggressively
        ratio = img.width / img.height
        target_h = int(usable_h * 0.92)
        target_w = int(target_h * ratio)
        if target_w > usable_w:
            target_w = usable_w
            target_h = int(target_w / ratio)

        img = img.resize((target_w, target_h))
        x = left_w + col_gap + pad + (usable_w - target_w)//2
        y = header_h + pad*2
        # soft border
        draw.rounded_rectangle([x-6, y-6, x+target_w+6, y+target_h+6], radius=18,
                               fill=WHITE, outline=(210,210,210), width=2)
        page.paste(img, (x, y))

    # --- footer
    footer_text = (
        f"{university}  •  {course}  •  "
        f"Generated: {datetime.now(timezone.utc).strftime('%Y-%m-%d %H:%M UTC')}"
    )
    draw.text((pad, H-40), footer_text, fill=(100,100,100), font=small_font)

    # --- save & return
    page.save(out_pdf, "PDF", resolution=300.0)
    return out_pdf
```
